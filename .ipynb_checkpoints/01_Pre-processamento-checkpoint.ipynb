# 2. CONJUNTO DE DADOS

```python
df = pd.read_csv('dados/lending_club_2007_2014.csv', index_col=0)

print(f'Linhas: {df.shape[0]:,}')
print(f'Colunas: {df.shape[1]}')
df.head()
```

# 3. ORDENAÇÃO POR ORDEM CRONOLÓGICA

```python
# issue_d vem como "Dec-11" (ano com 2 dígitos), então usamos %b-%y.
df['issue_d'] = pd.to_datetime(df['issue_d'], format='%b-%y')

# Ordena cronologicamente pela data de emissão
df = df.sort_values(by='issue_d', ascending=True).reset_index(drop=True)

# Confere o intervalo de datas
print(f"Data mais antiga: {df['issue_d'].min()}")
print(f"Data mais recente: {df['issue_d'].max()}")
df[['issue_d', 'id']].head()
```

# 4. VALORES FALTANTES

```python
missing_df = (
    df.isna().sum()
    .to_frame()
    .rename(columns={0: 'count'})
    .sort_values(by='count', ascending=False)
)
missing_df['pct'] = round(missing_df['count'] / df.shape[0] * 100, 3)
missing_df
```

```python
# Verificação de duplicados
print(f'Linhas totalmente duplicadas: {df.duplicated().sum()}')
print(f'IDs duplicados: {df["id"].duplicated().sum()}')
```

```python
# Colunas com 70% ou mais de valores ausentes (candidatas a remoção)
limite_ausencia = 70
cols_alta_ausencia = missing_df[missing_df['pct'] >= limite_ausencia].index.tolist()

print(f'Colunas com >= {limite_ausencia}% de ausência: {len(cols_alta_ausencia)}')
cols_alta_ausencia
```

### Remoção de variáveis com alta ausência

Estabelecemos um corte de 70%: variáveis com 70% ou mais de valores ausentes são removidas, pois a imputação em volume tão alto introduziria viés. As 20 colunas identificadas se enquadram em quatro motivos:

- **Variáveis de co-tomador** (`annual_inc_joint`, `dti_joint`, `verification_status_joint`): preenchidas apenas em aplicações conjuntas, que são minoria. Ausência estrutural.
- **Variáveis de linhas de crédito parceladas e rotativas** (`open_acc_6m`, `open_il_6m`, `open_il_12m`, `open_il_24m`, `mths_since_rcnt_il`, `total_bal_il`, `il_util`, `open_rv_12m`, `open_rv_24m`, `max_bal_bc`, `all_util`, `inq_fi`, `total_cu_tl`, `inq_last_12m`): passaram a ser coletadas apenas em parte do período, ficando vazias nos empréstimos anteriores. Ausência por disponibilidade temporal.
- **Variáveis de tempo desde o último evento** (`mths_since_last_record`, `mths_since_last_major_derog`): vazias quando o evento nunca ocorreu. Embora o vazio tenha significado, a proporção de ausência é alta demais para tratá-las como categoria.
- **Texto livre** (`desc`): descrição não estruturada preenchida pelo tomador, sem utilidade para o modelo.

```python
# Remoção das colunas com ausência >= 70%
df = df.drop(columns=cols_alta_ausencia)

print(f'Colunas removidas: {len(cols_alta_ausencia)}')
print(f'Shape atual: {df.shape}')
```

### Investigação dos valores ausentes remanescentes

Removidas as variáveis com alta ausência, investigamos as que restaram para decidir o tratamento. A ausência não é tratada de forma única: seu significado depende da causa. Analisamos a distribuição de `loan_status` nas linhas com ausência (para verificar se o vazio está associado a mais risco) e a natureza de cada variável, classificando cada caso em remover a observação, imputar um valor com significado ou tratar o vazio como categoria própria.

```python
# Ausentes remanescentes (colunas que ainda têm algum vazio)
missing_restante = (
    df.isna().sum()
    .to_frame(name='count')
    .query('count > 0')
    .sort_values('count', ascending=False)
)
missing_restante['pct'] = round(missing_restante['count'] / df.shape[0] * 100, 3)
missing_restante
```

```python
# emp_length: a ausência está associada a mais risco?
print('--- loan_status onde emp_length é ausente ---')
print(df.loc[df['emp_length'].isna(), 'loan_status'].value_counts(normalize=True) * 100)
```

### Tratamento dos valores ausentes remanescentes

Aplicamos tratamentos distintos conforme a natureza da ausência. As variáveis com pouquíssimos ausentes (`annual_inc`, `total_acc`, `acc_now_delinq`, `pub_rec`, `open_acc`, `delinq_2yrs`, `earliest_cr_line`, `inq_last_6mths`, `revol_util` e `collections_12_mths_ex_med`) têm suas observações removidas; dado o tamanho da base, essa remoção não introduz viés relevante. Para `emp_length`, cuja ausência se mostrou não associada a maior risco de default, imputamos "0 years", interpretando o vazio como ausência de experiência profissional formal.

As variáveis com ausência elevada e informativa (`mths_since_last_delinq`, `tot_coll_amt`, `tot_cur_bal` e `total_rev_hi_lim`) não são imputadas nesta etapa: seus valores ausentes serão tratados como categoria própria durante a discretização (binning) na etapa de WoE, preservando a informação contida na ausência.

```python
# Remoção das observações com ausência ínfima (29 registros no total)
cols_dropna = ['annual_inc', 'total_acc', 'acc_now_delinq', 'pub_rec',
               'open_acc', 'delinq_2yrs', 'earliest_cr_line', 'inq_last_6mths','revol_util', 'collections_12_mths_ex_med']
df = df.dropna(subset=cols_dropna)

# Imputação de emp_length: vazio interpretado como sem experiência formal
df['emp_length'] = df['emp_length'].fillna('0 years')

print(f'Shape após tratamento de ausentes: {df.shape}')
```

# 5. REMOÇÃO DE COLUNAS IRRELEVANTES E DE VAZAMENTO

### Remoção de colunas irrelevantes e de vazamento

Além das variáveis com alta ausência, removemos dois grupos. O primeiro reúne identificadores e campos sem valor preditivo: `id`, `member_id`, `url`, `zip_code`, `title`, `emp_title`, `application_type`, `pymnt_plan` e `policy_code`. O segundo reúne variáveis que só passam a existir após a concessão do empréstimo e que, portanto, constituiriam vazamento de informação: `next_pymnt_d`, `last_pymnt_d`, `last_credit_pull_d`, `funded_amnt_inv`, `total_pymnt_inv` e `out_prncp_inv`.

```python
# Colunas irrelevantes (identificadores, texto livre, baixa variância)
cols_irrelevantes = [
    'id', 'member_id', 'url', 'zip_code', 'title', 'emp_title',
    'application_type', 'pymnt_plan', 'policy_code'
]

# Colunas de vazamento (só existem após a concessão do empréstimo)
cols_vazamento = [
    'next_pymnt_d', 'last_pymnt_d', 'last_credit_pull_d',
    'funded_amnt_inv', 'total_pymnt_inv', 'out_prncp_inv'
]

df = df.drop(columns=cols_irrelevantes + cols_vazamento)
print(f'Colunas removidas nesta etapa: {len(cols_irrelevantes + cols_vazamento)}')
print(f'Shape atual: {df.shape}')
```

# 6. CONVERSÃO DE TIPOS E EXTRAÇÃO DE DATA

```python
# Inspeção dos formatos antes de converter
print('term:', df['term'].unique())
print()
print('emp_length:', df['emp_length'].unique())
print()
print('earliest_cr_line:', df['earliest_cr_line'].head(3).tolist())
```

### Conversão de tipos

Três variáveis estão como texto e precisam ser convertidas para uso no modelo:

- **`term`**: vem como `' 36 months'` / `' 60 months'`. Extraímos o número e convertemos para inteiro.
- **`emp_length`**: vem em faixas textuais (`'< 1 year'` a `'10+ years'`). Convertemos para inteiro, tratando `'< 1 year'` como 0 e `'10+ years'` como 10, o que preserva a ordem da variável.
- **`earliest_cr_line`**: data da primeira linha de crédito, no formato `'Oct-03'`. Convertemos para data.

```python
# term: extrai o número de meses e converte para inteiro
df['term'] = df['term'].str.extract(r'(\d+)').astype('int')
print('term:', df['term'].unique(), '| tipo:', df['term'].dtype)
```

```python
# emp_length: converte faixas textuais em inteiro (0 a 10)
df['emp_length'] = (
    df['emp_length']
    .str.replace('< 1 year', '0', regex=False)
    .str.replace('10+ years', '10', regex=False)
    .str.replace('+', '', regex=False)
    .str.replace(' years', '', regex=False)
    .str.replace(' year', '', regex=False)
    .astype('int')
)
print('emp_length:', sorted(df['emp_length'].unique()), '| tipo:', df['emp_length'].dtype)
```

```python
# earliest_cr_line: converte para data (formato "Oct-03")
df['earliest_cr_line'] = pd.to_datetime(df['earliest_cr_line'], format='%b-%y')
print(df['earliest_cr_line'].min(), '→', df['earliest_cr_line'].max())
```

### Criação de variável e correção de datas inconsistentes

Extraímos de `earliest_cr_line` a variável `mths_since_earliest_cr_line`, que mede há quantos meses o tomador abriu sua primeira linha de crédito, uma proxy de maturidade do histórico de crédito. Adotamos dezembro de 2017 como data de referência.

A conversão de datas com ano de dois dígitos gerou uma inconsistência: anos antigos (ex.: 1968) foram interpretados como futuros (2068), produzindo valores negativos de meses. Como esses casos correspondem a linhas de crédito muito antigas, imputamos o valor máximo observado da variável, preservando o significado de "primeira linha de crédito há muito tempo".

```python
# Data de referência para o cálculo
DATA_REF = pd.to_datetime('2017-12-01')

# Meses desde a primeira linha de crédito, calculado pela diferença de calendário
# (evita a ambiguidade de "mês" como duração fixa)
df['mths_since_earliest_cr_line'] = (
    (DATA_REF.year - df['earliest_cr_line'].dt.year) * 12
    + (DATA_REF.month - df['earliest_cr_line'].dt.month)
)

# Diagnóstico: quantos valores ficaram negativos (efeito do ano de 2 dígitos)
n_neg = (df['mths_since_earliest_cr_line'] < 0).sum()
print(f'Valores negativos (datas mal interpretadas): {n_neg}')
print(df['mths_since_earliest_cr_line'].describe())
```

### Correção dos valores negativos

Os 1.169 valores negativos resultam de datas antigas (anos 1960) interpretadas como futuras pela conversão de ano com dois dígitos. Como esses casos representam, na realidade, tomadores com histórico de crédito muito antigo, imputamos o valor máximo observado da variável, preservando o significado de "primeira linha de crédito estabelecida há muito tempo".

```python
# Imputa o máximo observado nos casos negativos (crédito muito antigo)
maximo_valido = df['mths_since_earliest_cr_line'].max()
df.loc[df['mths_since_earliest_cr_line'] < 0, 'mths_since_earliest_cr_line'] = maximo_valido

# Confirma que não há mais negativos
print(f'Valores negativos após correção: {(df["mths_since_earliest_cr_line"] < 0).sum()}')
print(df['mths_since_earliest_cr_line'].describe())
```

### Remoção de variável de data já processada

Extraída a variável `mths_since_earliest_cr_line`, a data original `earliest_cr_line` não é mais necessária e é removida. Mantemos `issue_d`, que ainda será utilizada no split temporal (out-of-time) da modelagem.

```python
# Remove a data original, já resumida em mths_since_earliest_cr_line
df = df.drop(columns=['earliest_cr_line'])
print(f'Shape atual: {df.shape}')
```

# 7. VARIÁVEL-ALVO E REMOÇÃO DE VAZAMENTO PÓS-CONCESSÃO

```python
# Categorias de loan_status antes de definir o alvo
df['loan_status'].value_counts()
```

### Definição da variável-alvo (default)

Antes de criar o alvo, descartamos os contratos em andamento (`Current`), cujo desfecho ainda é desconhecido e que, portanto, não podem ser rotulados como bons ou maus pagadores. Em seguida, criamos a variável binária `default`, atribuindo `0` aos inadimplentes (`Charged Off`, `Late (31-120 days)`, `Default` e `Does not meet the credit policy. Status:Charged Off`) e `1` aos adimplentes. Nesta convenção, `1` representa o bom pagador, o que faz com que coeficientes positivos no modelo estejam associados a menor risco.

```python
# Descarta contratos em andamento (desfecho desconhecido)
df = df[df['loan_status'] != 'Current'].copy()

# Alvo do modelo de PD: 1 = adimplente (bom pagador), 0 = inadimplente
inadimplentes = ['Charged Off', 'Late (31-120 days)',
                 'Does not meet the credit policy. Status:Charged Off', 'Default']
df['default'] = np.where(df['loan_status'].isin(inadimplentes), 0, 1)

print(f'Shape após remover Current: {df.shape}')
print()
print('Distribuição do alvo (%):')
print(df['default'].value_counts(normalize=True) * 100)
```

```python
df.info()
```

### Remoção de variáveis com vazamento de informação

Nesta etapa removemos todas as variáveis que só passam a existir após a concessão do empréstimo e que, portanto, não estariam disponíveis no momento da decisão de crédito. Incluí-las caracterizaria vazamento de informação, inflando artificialmente o desempenho do modelo. São elas: saldo e pagamentos do contrato (`out_prncp`, `total_pymnt`, `total_rec_prncp`, `total_rec_int`, `total_rec_late_fee`, `last_pymnt_amnt`) e valores de recuperação pós-inadimplência (`recoveries`, `collection_recovery_fee`). Removemos também `loan_status`, já resumida na variável-alvo `default`.

```python
# Colunas com vazamento: só existem após a concessão do empréstimo
cols_vazamento_pos = [
    'out_prncp', 'total_pymnt', 'total_rec_prncp', 'total_rec_int',
    'total_rec_late_fee', 'recoveries', 'collection_recovery_fee', 'last_pymnt_amnt'
]

# loan_status: origem do alvo, remover para não vazar o próprio target
df = df.drop(columns=cols_vazamento_pos + ['loan_status'])

print(f'Colunas de vazamento removidas: {len(cols_vazamento_pos) + 1}')
print(f'Shape atual: {df.shape}')
```

```python
# Ausentes restantes: quantidade e percentual
restantes = (
    df.isna().sum()
    .to_frame(name='qtd')
    .query('qtd > 0')
    .sort_values('qtd', ascending=False)
)
restantes['pct'] = round(restantes['qtd'] / df.shape[0] * 100, 2)
restantes
```

### Tratamento das ausências informativas

As variáveis `mths_since_last_delinq`, `tot_coll_amt`, `tot_cur_bal` e `total_rev_hi_lim` apresentam ausência elevada e informativa (associada ao risco). Em vez de imputá-las de forma cega, o que apagaria o sinal contido na ausência, criamos para cada uma uma variável indicadora (flag) que registra a ocorrência do valor ausente, e em seguida preenchemos o valor original. Assim, a informação "havia ausência" é preservada explicitamente e o conjunto de dados fica sem valores nulos. Para `mths_since_last_delinq`, cujo vazio significa ausência de inadimplência anterior, o preenchimento usa um valor alto, coerente com "sem inadimplência recente".

```python
# Colunas com ausência informativa
cols_flag = ['mths_since_last_delinq', 'tot_coll_amt', 'tot_cur_bal', 'total_rev_hi_lim']

# 1) Cria a flag de ausência para cada uma (1 = era ausente, 0 = tinha valor)
for col in cols_flag:
    df[f'{col}_missing'] = df[col].isna().astype('int8')

# 2) Preenche os valores originais
# mths_since_last_delinq: vazio = nunca teve inadimplência -> valor alto
valor_alto = df['mths_since_last_delinq'].max()
df['mths_since_last_delinq'] = df['mths_since_last_delinq'].fillna(valor_alto)

# tot_coll_amt, tot_cur_bal, total_rev_hi_lim: preenche com 0
for col in ['tot_coll_amt', 'tot_cur_bal', 'total_rev_hi_lim']:
    df[col] = df[col].fillna(0)

# Confirmação: não deve restar nenhum ausente
print(f'Total de ausentes no dataframe: {df.isna().sum().sum()}')
print(f'Shape final: {df.shape}')
```

# 8. ENGENHARIA DE ATRIBUTOS

Criamos duas variáveis derivadas com significado econômico direto na análise de crédito, ambas medindo a capacidade de pagamento do tomador em relação à sua renda:

- **`parcela_renda`** (`installment` / renda mensal): proporção da renda mensal comprometida com a parcela do empréstimo. Valores altos indicam maior comprometimento e, potencialmente, maior risco.
- **`emprestimo_renda`** (`loan_amnt` / `annual_inc`): tamanho do empréstimo em relação à renda anual. Mede o grau de alavancagem do tomador.

Essas razões capturam informação que os valores absolutos isolados não expressam, e serão avaliadas junto às demais variáveis na etapa de modelagem.

```python
# Razão parcela/renda: comprometimento da renda mensal com a parcela
df['parcela_renda'] = df['installment'] / (df['annual_inc'] / 12)

# Razão empréstimo/renda: alavancagem do tomador
df['emprestimo_renda'] = df['loan_amnt'] / df['annual_inc']

print(df[['parcela_renda', 'emprestimo_renda']].describe().round(4))
print(f'\nShape atual: {df.shape}')
```
