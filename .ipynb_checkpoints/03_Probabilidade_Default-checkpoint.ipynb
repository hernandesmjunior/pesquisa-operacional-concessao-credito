# 3. DIVISÃO TREINO-TESTE (OUT-OF-TIME)

Em problemas de crédito, a divisão entre treino e teste é feita de forma temporal (*out-of-time*), e não aleatória. O modelo é treinado com os empréstimos mais antigos e avaliado nos mais recentes, reproduzindo a situação real de uso: prever o comportamento de tomadores futuros a partir de dados do passado. Como os dados já estão ordenados cronologicamente por `issue_d`, utilizamos os primeiros 70% das observações para treino e os 30% finais para teste.

```python
# Divisão out-of-time: 70% mais antigos para treino, 30% mais recentes para teste
# (o dataframe já está ordenado por issue_d, feito no notebook 01)
ponto_corte = int(len(df) * 0.70)

train = df.iloc[:ponto_corte].copy()
test = df.iloc[ponto_corte:].copy()

print(f'Treino: {train.shape[0]:,} linhas | {train["issue_d"].min().date()} a {train["issue_d"].max().date()}')
print(f'Teste:  {test.shape[0]:,} linhas | {test["issue_d"].min().date()} a {test["issue_d"].max().date()}')
```

```python
# Distribuição do alvo em treino e teste (1 = mau pagador, 0 = bom pagador)
dist_train = train['default'].value_counts(normalize=True).sort_index() * 100
dist_test = test['default'].value_counts(normalize=True).sort_index() * 100
comp = pd.DataFrame({'treino (%)': dist_train.round(2), 'teste (%)': dist_test.round(2)})
comp.index = ['Bom pagador (0)', 'Mau pagador (1)']
print(comp)
print()
print(f'Taxa de inadimplência no treino: {(train["default"]==1).mean()*100:.2f}%')
print(f'Taxa de inadimplência no teste:  {(test["default"]==1).mean()*100:.2f}%')
```

```python
# Taxa de inadimplência por mês de emissão
inad_mensal = (df.groupby('issue_d')
    .agg(total=('default','size'), inadimplentes=('default','sum')))
inad_mensal['taxa_inad'] = inad_mensal['inadimplentes']/inad_mensal['total']*100
data_corte = train['issue_d'].max()
print(f'Fronteira treino/teste: {data_corte.date()}')
inad_mensal.tail(12)
```

```python
# Gráfico: taxa de default mensal com a fronteira treino/teste
fig, ax = plt.subplots(figsize=(14, 5))

ax.plot(inad_mensal.index, inad_mensal['taxa_inad'],
        marker='o', markersize=3, color='#4c72b0', label='Taxa de default')
ax.axvline(data_corte, color='#d9534f', linestyle='--', label='Fronteira treino/teste')

ax.set_title('Taxa de default por mês de emissão')
ax.set_xlabel('')
ax.set_ylabel('Taxa de default (%)')
ax.legend()
plt.tight_layout()
plt.show()
```

### Diferença de inadimplência entre treino e teste

A divisão out-of-time produz treino com 19,33% de inadimplência e teste com 26,35% — agora o teste é o **mais alto**, e não o mais baixo. Esse é o oposto do padrão da base que mantinha os contratos em andamento, e decorre da definição de alvo adotada no notebook 01, que retém apenas desfechos resolvidos.

Entre as safras mais recentes, os contratos que já alcançaram um desfecho são, em proporção, os que inadimpliram cedo; os bons pagadores recentes em geral ainda estariam em andamento (`Current`) e foram excluídos da amostra. Esse efeito de seleção eleva a inadimplência observada no período de teste. É o reverso, bem mais brando, do viés de maturação que a exclusão dos contratos em andamento eliminou.

A diferença não compromete a capacidade discriminatória do modelo: as métricas de ordenação (KS e AUC) medem o poder de separar bons e maus pagadores, robusto a variações na taxa base, e serão avaliadas na Seção 8.

# 4. WEIGHT OF EVIDENCE (WoE) E INFORMATION VALUE (IV)

Antes de estimar o modelo, as variáveis são transformadas pela técnica de *Weight of Evidence* (WoE), amplamente utilizada na construção de *scorecards* de crédito. A ideia central é discretizar cada variável em faixas (categorias) e substituir cada faixa por um valor que expressa o quanto ela separa bons de maus pagadores.

## Weight of Evidence (WoE)

Para cada faixa de uma variável, o WoE é definido como:

$$WoE = \ln\left(\frac{\text{proporção de bons pagadores na faixa}}{\text{proporção de maus pagadores na faixa}}\right)$$

Onde a "proporção de bons" é o número de bons pagadores da faixa dividido pelo total de bons pagadores, e analogamente para os maus. A interpretação é direta:

- **WoE positivo**: a faixa concentra proporcionalmente mais bons pagadores que maus (menor risco).
- **WoE negativo**: a faixa concentra proporcionalmente mais maus pagadores (maior risco).
- **WoE próximo de zero**: a faixa não discrimina risco.

A transformação WoE traz vantagens para a regressão logística: lida naturalmente com variáveis categóricas e com valores ausentes (que se tornam uma faixa própria), acomoda relações não lineares entre a variável e o risco, e coloca todas as variáveis em uma escala comparável.

## Information Value (IV)

Enquanto o WoE mede o poder discriminatório de cada *faixa*, o *Information Value* (IV) resume o poder preditivo da *variável inteira* em um único número:

$$IV = \sum_{\text{faixas}} (\text{prop. de bons} - \text{prop. de maus}) \times WoE$$

O IV é usado para selecionar variáveis. Uma referência usual de interpretação é:

| IV | Poder preditivo |
|---|---|
| < 0,02 | Sem poder preditivo |
| 0,02 – 0,10 | Fraco |
| 0,10 – 0,30 | Médio |
| 0,30 – 0,50 | Forte |
| > 0,50 | Muito forte (verificar sobreajuste) |

Variáveis com IV muito baixo são descartadas por não contribuírem para o modelo; variáveis com IV muito alto devem ser avaliadas com cautela, pois um poder preditivo excessivo pode sinalizar sobreajuste ou vazamento de informação.

Para ilustrar o cálculo, aplicamos a técnica a uma variável categórica (`grade`), que não requer discretização prévia. Os valores são calculados **apenas sobre o conjunto de treino**, evitando vazamento de informação do teste.

```python
def calcular_woe_iv(dados, feature, alvo='default'):
    """
    Calcula WoE por faixa e IV total de uma variável.
    Convenção do alvo: 1 = mau pagador (inadimplente), 0 = bom pagador.
    WoE = ln(prop_bons / prop_maus).
    """
    g = dados.groupby(feature, observed=True)[alvo].agg(['count', 'sum'])
    g.columns = ['total', 'maus']              # soma do alvo = nº de maus (alvo=1)
    g['bons'] = g['total'] - g['maus']
    g['prop_bons'] = g['bons'] / g['bons'].sum()
    g['prop_maus'] = g['maus'] / g['maus'].sum()
    g['woe'] = np.log(g['prop_bons'] / g['prop_maus'])
    g['iv_faixa'] = (g['prop_bons'] - g['prop_maus']) * g['woe']
    g['taxa_default_%'] = g['maus'] / g['total'] * 100
    iv = g['iv_faixa'].sum()
    return g.round(4), round(iv, 4)

tabela_grade, iv_grade = calcular_woe_iv(train, 'grade')
print(tabela_grade[['total', 'bons', 'maus', 'taxa_default_%', 'woe', 'iv_faixa']])
print(f'\nIV de grade: {iv_grade}')
```

```python
# Visualização do WoE por faixa de grade
plt.figure(figsize=(9, 4))
plt.bar(tabela_grade.index.astype(str), tabela_grade['woe'],
        color=['#5cb85c' if w > 0 else '#d9534f' for w in tabela_grade['woe']])
plt.axhline(0, color='black', linewidth=0.8)
plt.title('WoE por faixa de grade')
plt.xlabel('grade')
plt.ylabel('WoE')
plt.tight_layout()
plt.show()
```

O `grade` apresenta WoE perfeitamente monotônico: decresce de +1,17 (grade A) a -1,12 (grade G), acompanhando o aumento da taxa de inadimplência, que vai de 6,9% a 42,3%. Com IV de 0,369, situa-se na faixa de poder preditivo forte (0,30 a 0,50). Esse comportamento confirma que a classificação de risco do Lending Club ordena adequadamente os tomadores.

# 5. SELEÇÃO DE VARIÁVEIS POR INFORMATION VALUE

Antes de discretizar as variáveis com detalhe, calculamos o Information Value (IV) de cada candidata para identificar as mais preditivas. Esse passo evita investir esforço em variáveis sem poder discriminatório. A variável `addr_state` (estado do tomador) foi excluída por decisão metodológica: o uso da localização geográfica na concessão de crédito pode introduzir viés discriminatório contra determinadas regiões, o que é indesejável em um modelo responsável.

```python
# Variáveis candidatas (addr_state excluída por risco de viés geográfico)
alvo = 'default'
excluir = [alvo, 'issue_d', 'addr_state']

categoricas = [c for c in train.select_dtypes(include=['object', 'category', 'string']).columns
               if c not in excluir]
numericas = [c for c in train.select_dtypes(include=[np.number]).columns
             if c not in excluir]

# IV de cada variável (numéricas discretizadas por decis, categóricas diretas)
resultados = []
tmp = train.copy()
for col in numericas:
    try:
        tmp['_bin'] = pd.qcut(train[col], q=10, duplicates='drop')
        _, iv = calcular_woe_iv(tmp, '_bin')
    except Exception:
        iv = np.nan
    resultados.append(('numérica', col, iv))
for col in categoricas:
    _, iv = calcular_woe_iv(train, col)
    resultados.append(('categórica', col, iv))

tabela_iv = (pd.DataFrame(resultados, columns=['tipo', 'variavel', 'IV'])
             .sort_values('IV', ascending=False)
             .reset_index(drop=True))
tabela_iv
```

Aplicamos um corte de IV ≥ 0,02 para selecionar as variáveis. Esse limiar corresponde à fronteira de "sem poder preditivo" adotada na Seção 4, descartando variáveis que praticamente não discriminam risco, incluindo as indicadoras de ausência (cujo IV resultou nulo, pois o sinal da ausência já é capturado pela discretização das próprias variáveis) e as variáveis quase constantes identificadas na análise exploratória.

```python
# Corte de seleção: IV >= 0.02
LIMIAR_IV = 0.02
selecionadas = tabela_iv[tabela_iv['IV'] >= LIMIAR_IV]['variavel'].tolist()

# Inclusão manual por relevância de domínio (IV abaixo de 0,02 nesta base, mas
# preditores clássicos em risco de crédito): inq_last_6mths (consultas recentes de
# crédito) e home_ownership (situação de moradia).
for v in ['inq_last_6mths', 'home_ownership']:
    if v not in selecionadas:
        selecionadas.append(v)

print(f'Variáveis selecionadas ({len(selecionadas)}):')
for v in selecionadas:
    print(f'  {v}')
```

O corte seleciona as variáveis com IV ≥ 0,02. A elas acrescentamos manualmente, por relevância de domínio, duas variáveis cujo IV ficou logo abaixo do limiar nesta base: `inq_last_6mths` (0,018), número de consultas de crédito nos últimos seis meses, indicador clássico de busca por crédito em múltiplas instituições; e `home_ownership` (0,009), situação de moradia do tomador. Ambas são preditores reconhecidos em risco de crédito e são mantidas no modelo. A variável `verification_status` (IV 0,040) entra naturalmente pelo corte.

### Tratamento de redundância: variáveis de risco

As variáveis `grade`, `sub_grade` e `int_rate` representam, essencialmente, a mesma informação: o Lending Club atribui uma classificação de risco (`grade`), refina-a em subclasses (`sub_grade`) e define a taxa de juros (`int_rate`) a partir dessa classificação. Isso se reflete tanto na forte correlação observada na análise exploratória quanto nos valores de IV muito próximos das três (entre 0,37 e 0,40). Manter as três no modelo introduziria multicolinearidade sem agregar poder preditivo.

Optou-se por manter apenas `grade`, descartando `sub_grade` e `int_rate`, pelos seguintes motivos: `grade` possui apenas sete categorias já ordenadas por risco, o que dispensa agrupamento adicional e favorece a interpretabilidade do scorecard; seu IV é praticamente equivalente ao das demais; e, conceitualmente, representa a avaliação de risco original da instituição, da qual as outras duas derivam.

```python
# Remove sub_grade e int_rate: redundantes com grade (mesma informação de risco)
for v in ['sub_grade', 'int_rate']:
    if v in selecionadas:
        selecionadas.remove(v)

print(f'Após tratar redundância de risco ({len(selecionadas)} variáveis):')
for v in selecionadas:
    print(f'  {v}')
```

### Tratamento de redundância: variáveis de valor

As variáveis `loan_amnt`, `funded_amnt` e `installment` são altamente correlacionadas entre si (acima de 0,95, conforme a análise exploratória), pois representam o valor do empréstimo e sua parcela, que dele deriva. Além disso, a informação de valor foi incorporada de forma mais rica às variáveis criadas na engenharia de atributos: `emprestimo_renda` (derivada de `loan_amnt`) e `parcela_renda` (derivada de `installment`), que relativizam o valor pela renda do tomador.

Essas razões apresentam poder preditivo bem maior que as variáveis de valor brutas das quais derivam: `emprestimo_renda` tem IV de 0,143 e `parcela_renda`, 0,111, contra 0,02 a 0,04 de `loan_amnt`, `funded_amnt` e `installment`. Optou-se, portanto, por descartar as três variáveis de valor brutas, mantendo apenas as razões, que capturam a mesma informação de forma mais discriminante e sem redundância.

```python
# Remove as variáveis de valor brutas: redundantes entre si e já incorporadas
# de forma relativizada em emprestimo_renda e parcela_renda
for v in ['funded_amnt', 'loan_amnt', 'installment']:
    if v in selecionadas:
        selecionadas.remove(v)

print(f'Variáveis finais selecionadas ({len(selecionadas)}):')
for v in selecionadas:
    print(f'  {v}')
```

# 6. DISCRETIZAÇÃO E CÁLCULO DE WoE

As variáveis numéricas selecionadas são discretizadas em faixas por decis (quantis de igual frequência) e, em seguida, transformadas em WoE. Para cada uma, verificamos se o WoE resultante é monotônico. As variáveis cuja não monotonicidade é estrutural, causada pelo acúmulo de zeros das ausências imputadas no pré-processamento (`tot_cur_bal` e `total_rev_hi_lim`), recebem ajuste manual, com o valor zero isolado em faixa própria. As demais variáveis contínuas resultam monotônicas já na discretização por decis, dispensando reagrupamento — e, de todo modo, a própria transformação WoE acomoda relações não lineares entre a variável e o risco. As variáveis categóricas não requerem discretização.

```python
def eh_monotonica(woes):
    d = np.diff(woes)
    return bool(np.all(d <= 0) or np.all(d >= 0))

# Numéricas selecionadas
num_selecionadas = ['emprestimo_renda', 'parcela_renda', 'dti', 'revol_util',
                    'annual_inc', 'tot_cur_bal', 'total_rev_hi_lim', 'inq_last_6mths']

# Diagnóstico: discretiza por decis e verifica monotonicidade
print(f'{"variável":<20}{"n_bins":<8}{"IV":<10}{"monotônica"}')
print('-' * 50)
for col in num_selecionadas:
    tmp = train.copy()
    tmp['_bin'] = pd.qcut(train[col], q=10, duplicates='drop')
    tab, iv = calcular_woe_iv(tmp, '_bin')
    print(f'{col:<20}{len(tab):<8}{iv:<10}{eh_monotonica(tab["woe"].values)}')
```

### Ajuste de `tot_cur_bal`

A variável `tot_cur_bal` apresenta cerca de 35% de valores iguais a zero (ausências preenchidas no pré-processamento) e WoE não monotônico na discretização por decis. O valor zero representa ausência de informação, e é isolado em faixa própria. Nesta base, a faixa "zero" tem WoE positivo (+0,20), ou seja, risco relativo menor: sua taxa de inadimplência (16,4%) fica abaixo da média do treino (19,3%). Entre as faixas com valor, o WoE é monotônico, com o risco caindo conforme o saldo aumenta. Vale notar que na base que mantinha os contratos em andamento a faixa zero era de maior risco; a inversão decorre da composição mais antiga da amostra de desfechos resolvidos, em que a ausência desse campo é mais frequente.

```python
# tot_cur_bal: zero isolado (ausência) + faixas monotônicas (quintis agrupados)
bins_tcb = [-1, 0, 22569, 131376, 236223, np.inf]
labels_tcb = ['zero', 'ate_22k', '22k_131k', '131k_236k', '236k+']
train['tot_cur_bal_bin'] = pd.cut(train['tot_cur_bal'], bins=bins_tcb, labels=labels_tcb)

tab_tcb, iv_tcb = calcular_woe_iv(train, 'tot_cur_bal_bin')
print(tab_tcb[['total', 'maus', 'woe']].round(4).reindex(labels_tcb))
print(f'\nIV: {iv_tcb}')
```

### Ajuste de `total_rev_hi_lim`

Assim como `tot_cur_bal`, a variável `total_rev_hi_lim` tem cerca de 35% de zeros e WoE não monotônico por decis. Aplicamos o mesmo tratamento: o zero é isolado em faixa própria e o restante é discretizado em quintis. A faixa "zero" novamente apresenta WoE positivo (menor risco), e as faixas com valor são monotônicas, com o risco caindo conforme o limite de crédito rotativo aumenta.

```python
# total_rev_hi_lim: zero isolado + quintis monotônicos para o restante
bins_trhl = [-1, 0, 11700, 18300, 26600, 40600, np.inf]
labels_trhl = ['zero', 'ate_11.7k', '11.7k_18.3k', '18.3k_26.6k', '26.6k_40.6k', '40.6k+']
train['total_rev_hi_lim_bin'] = pd.cut(train['total_rev_hi_lim'], bins=bins_trhl, labels=labels_trhl)

tab_trhl, iv_trhl = calcular_woe_iv(train, 'total_rev_hi_lim_bin')
print(tab_trhl[['total', 'maus', 'woe']].round(4).reindex(labels_trhl))
print(f'\nIV: {iv_trhl}')
```

### Transformação WoE de treino e teste

Consolidamos a discretização e a transformação WoE das 12 variáveis selecionadas. As cinco contínuas são discretizadas por decis; `tot_cur_bal`, `total_rev_hi_lim` e `inq_last_6mths` usam os cortes manuais definidos anteriormente; as quatro categóricas dispensam discretização.

Ponto central para evitar vazamento: os cortes e os valores de WoE são aprendidos **apenas no treino** e depois aplicados ao teste, sem qualquer reestimação. As bordas das faixas são estendidas ao infinito para garantir que valores do teste fora do intervalo de treino ainda sejam alocados corretamente.

```python
def woe_map(tr, col_bin, alvo='default'):
    g = tr.groupby(col_bin, observed=True)[alvo].agg(['count', 'sum'])
    g.columns = ['t', 'maus']            # soma do alvo = maus (1=mau)
    g['bons'] = g['t'] - g['maus']
    pb = g['bons'] / g['bons'].sum()
    pm = g['maus'] / g['maus'].sum()
    return (np.log(pb / pm)).to_dict()

# Numéricas por decis (bem comportadas)
num_qcut = ['emprestimo_renda', 'parcela_renda', 'dti', 'revol_util', 'annual_inc']

# Numéricas com cortes manuais (zeros isolados / variável discreta)
cortes_manuais = {
    'tot_cur_bal':      [-np.inf, 0, 22569, 131376, 236223, np.inf],
    'total_rev_hi_lim': [-np.inf, 0, 11700, 18300, 26600, 40600, np.inf],
    'inq_last_6mths':   [-np.inf, 0, 1, 2, np.inf],  # faixas: 0, 1, 2, 3+
}

categoricas = ['grade', 'purpose', 'home_ownership', 'verification_status']  # categóricas (IV>=0.02 + domínio)

# --- Aprende cortes e mapas WoE SOMENTE no treino ---
edges, mapas = {}, {}
for col in num_qcut:
    _, e = pd.qcut(train[col], q=10, duplicates='drop', retbins=True)
    e[0], e[-1] = -np.inf, np.inf          # estende bordas para cobrir o teste
    edges[col] = e
    train[col + '_bin'] = pd.cut(train[col], bins=e)
    mapas[col] = woe_map(train, col + '_bin')

for col, c in cortes_manuais.items():
    edges[col] = c
    train[col + '_bin'] = pd.cut(train[col], bins=c)
    mapas[col] = woe_map(train, col + '_bin')

for col in categoricas:
    mapas[col] = woe_map(train, col)

# --- Aplica ao teste usando os cortes/mapas do treino ---
todas_num = num_qcut + list(cortes_manuais)
for col in todas_num:
    test[col + '_bin'] = pd.cut(test[col], bins=edges[col])

def transformar_woe(dados):
    X = pd.DataFrame(index=dados.index)
    for col in todas_num:
        X[col] = dados[col + '_bin'].map(mapas[col])
    for col in categoricas:
        # categorias não vistas no treino (ex.: 'ANY' no teste) recebem WoE neutro (0)
        X[col] = dados[col].map(mapas[col]).fillna(0)
    return X

X_train = transformar_woe(train)
X_test = transformar_woe(test)
y_train = train['default']
y_test = test['default']

print(f'X_train: {X_train.shape} | NaN: {X_train.isna().sum().sum()}')
print(f'X_test:  {X_test.shape} | NaN: {X_test.isna().sum().sum()}')
```

# 7. REGRESSÃO LOGÍSTICA

Com as variáveis transformadas em WoE, estimamos o modelo de probabilidade de default por regressão logística. Utilizamos a biblioteca `statsmodels`, que fornece os p-valores dos coeficientes, permitindo avaliar a significância estatística de cada variável, prática usual na construção de scorecards. Como a variável-alvo adota a convenção `1 = mau pagador` e o WoE é positivo nas faixas de menor risco, espera-se que todos os coeficientes sejam negativos: um WoE mais alto, associado a um perfil melhor, reduz a probabilidade estimada de default.

```python
# Adiciona a constante (intercepto) exigida pelo statsmodels
X_train_sm = sm.add_constant(X_train)
X_test_sm = sm.add_constant(X_test)

# Estima a regressão logística
modelo = sm.Logit(y_train, X_train_sm).fit(disp=False)

# Resumo com coeficientes, p-valores e significância
print(modelo.summary())
```

### Tratamento de redundância entre as razões de renda

As razões `emprestimo_renda` e `parcela_renda` têm correlação de 0,95 entre si, pois a parcela é proporcional ao valor do empréstimo e ambas medem o mesmo comprometimento da renda. Essa multicolinearidade distorce os coeficientes: no modelo inicial, `parcela_renda` recebe coeficiente positivo (+0,91), sinal contrário ao esperado — com a transformação WoE, todos os coeficientes deveriam ser negativos —, enquanto `emprestimo_renda` fica fortemente negativo (−1,27). Removemos `parcela_renda`, de IV ligeiramente menor (0,111 contra 0,143), e o efeito se consolida em `emprestimo_renda`, cujo coeficiente passa a −0,55, com sinal e magnitude coerentes.

```python
# Remove parcela_renda (redundante com emprestimo_renda, correlação 0.95)
X_train_final = X_train.drop(columns=['parcela_renda'])
X_test_final = X_test.drop(columns=['parcela_renda'])

# Reajusta o modelo com 11 variáveis
X_train_sm = sm.add_constant(X_train_final)
X_test_sm = sm.add_constant(X_test_final)
modelo = sm.Logit(y_train, X_train_sm).fit(disp=False)
print(modelo.summary())
```

# 8. VALIDAÇÃO DO MODELO

Avaliamos o desempenho do modelo com as métricas usuais em risco de crédito, calculadas tanto no treino quanto no teste (out-of-time) para verificar generalização:

- **AUC-ROC**: área sob a curva ROC, mede a capacidade de ordenar bons e maus pagadores.
- **Gini**: derivado do AUC (Gini = 2·AUC − 1), métrica tradicional em crédito.
- **KS (Kolmogorov-Smirnov)**: máxima separação entre as distribuições acumuladas de bons e maus pagadores.

```python
# Probabilidades preditas de default (classe 1 = mau pagador)
proba_train = modelo.predict(X_train_sm)
proba_test = modelo.predict(X_test_sm)

def metricas(y_true, proba):
    auc = roc_auc_score(y_true, proba)
    gini = 2 * auc - 1
    # KS: máxima diferença entre as acumuladas
    fpr, tpr, _ = roc_curve(y_true, proba)
    ks = np.max(np.abs(tpr - fpr))
    return auc, gini, ks

auc_tr, gini_tr, ks_tr = metricas(y_train, proba_train)
auc_te, gini_te, ks_te = metricas(y_test, proba_test)

resultados = pd.DataFrame({
    'Treino': [auc_tr, gini_tr, ks_tr],
    'Teste': [auc_te, gini_te, ks_te]
}, index=['AUC', 'Gini', 'KS']).round(4)
print(resultados)
```

O modelo apresenta desempenho consistente entre treino e teste (AUC de 0,694 e 0,699; KS de 0,281 e 0,289), indicando boa capacidade de generalização e ausência de sobreajuste. Os valores situam-se em patamar adequado para modelos de risco de crédito, com KS entre 0,25 e 0,35. O desempenho ligeiramente superior no teste não é anômalo em divisões temporais e reforça a ausência de overfitting. A diferença de taxa base entre treino e teste não afeta essas métricas, pois AUC e KS avaliam a ordenação do risco, robusta a variações na taxa base.

```python
# Curva ROC (treino e teste)
fpr_tr, tpr_tr, _ = roc_curve(y_train, proba_train)
fpr_te, tpr_te, _ = roc_curve(y_test, proba_test)

plt.figure(figsize=(7, 7))
plt.plot(fpr_tr, tpr_tr, label=f'Treino (AUC = {auc_tr:.3f})', color='#4c72b0')
plt.plot(fpr_te, tpr_te, label=f'Teste (AUC = {auc_te:.3f})', color='#dd8452')
plt.plot([0, 1], [0, 1], 'k--', linewidth=0.8, label='Aleatório')
plt.xlabel('Taxa de falsos positivos')
plt.ylabel('Taxa de verdadeiros positivos')
plt.title('Curva ROC')
plt.legend()
plt.tight_layout()
plt.show()
```

# 9. SCORECARD

A probabilidade de default é convertida em um score de crédito, uma escala em que valores maiores indicam menor risco, mais intuitiva para a decisão de concessão. A transformação é linear sobre o logaritmo das odds (razão entre a probabilidade de bom e a de mau pagador), padrão na construção de scorecards:

$$Score = offset + fator \times \ln(odds)$$

O parâmetro PDO (Points to Double the Odds) define de quantos pontos a odds dobra, e a âncora (score base e odds base) posiciona a escala. Ancoramos o score na odds mediana do conjunto de treino, e não do teste, de modo que a escala reflita o perfil real dos tomadores sem contaminar a calibração com o período de validação.

```python
# Com alvo 1 = mau pagador, o modelo prediz a probabilidade de default (PD) diretamente.
PDO = 40
score_base = 500
fator = PDO / np.log(2)

# PD no treino e âncora da escala no treino (evita usar o teste para calibrar)
PD_train = modelo.predict(X_train_sm)
odds_bom_train = (1 - PD_train) / PD_train          # odds de ser BOM pagador
odds_base = np.median(odds_bom_train)
offset = score_base - fator * np.log(odds_base)

# Aplica ao teste
train['PD'] = PD_train                              # PD no treino (para calibrar o corte)
test['PD'] = modelo.predict(X_test_sm)              # probabilidade de default
odds_bom = (1 - test['PD']) / test['PD']
test['score'] = offset + fator * np.log(odds_bom)   # score alto = menor risco

print(f'Score: mín {test["score"].min():.0f} | média {test["score"].mean():.0f} | máx {test["score"].max():.0f}')

plt.figure(figsize=(9,4))
plt.hist(test['score'], bins=40, color='#4c72b0', edgecolor='white')
plt.title('Distribuição do score de crédito (teste)')
plt.xlabel('Score'); plt.ylabel('Frequência'); plt.tight_layout(); plt.show()
```

A distribuição do score concentra-se em torno de 491, com caudas menos densas nos extremos (mínimo 369, máximo 621). Esse comportamento é desejável em um scorecard: a maior parte dos tomadores situa-se em uma faixa intermediária de risco, com poucos casos nos extremos de altíssimo ou baixíssimo risco. Scores maiores correspondem a menor probabilidade de default.

# 10. DEFINIÇÃO DA ELEGIBILIDADE

O ponto de corte é uma decisão de negócio, **calibrada na amostra de desenvolvimento (treino)** e apenas aplicada ao teste para validação — escolher o corte olhando o teste contaminaria a avaliação out-of-sample. Em vez de maximizar a estatística KS, adotamos o critério da *strategy curve*: observamos, no treino, como a taxa de aprovação e a inadimplência acumulada da carteira aprovada variam conforme o corte, e escolhemos o maior corte cuja inadimplência acumulada dos aprovados permaneça em até 12% — teto de risco definido como política de concessão.

O corte resultante é **PD ≤ 20%**: no treino, aprova 58,3% dos contratos (96.838 de 166.145) com inadimplência acumulada de 11,86%, dentro do teto de 12%. São elegíveis os clientes com probabilidade de default de até 20%, que seguem para a etapa de otimização da carteira; os demais são reprovados nesta fase.

```python
# --- Ponto de corte de elegibilidade ---
# Calibrado no TREINO pela strategy curve: escolhemos o maior corte que mantém a
# inadimplência acumulada da carteira aprovada em até 12%. O corte resultante é
# PD <= 20%. Ele é apenas APLICADO ao teste para validação (não se escolhe pelo teste).
CORTE_RISCO = 20          # score de risco (PD em %) máximo para ser elegível

train['elegivel'] = (train['PD']*100 <= CORTE_RISCO).astype(int)
test['elegivel']  = (test['PD']*100  <= CORTE_RISCO).astype(int)

for nome, dfx in [('TREINO (calibração)', train), ('TESTE (validação)', test)]:
    el = dfx['elegivel'] == 1
    print(f'[{nome}] elegíveis {int(el.sum()):,} de {len(dfx):,} ({el.mean()*100:.1f}%) | '
          f'inad. aprovados {dfx.loc[el,"default"].mean()*100:.2f}% | base {dfx["default"].mean()*100:.2f}%')
print(f"\nInadimplência entre INELEGÍVEIS (teste): {test.loc[test['elegivel']==0,'default'].mean()*100:.2f}%")
```

```python
# Strategy curve no TREINO (onde o corte é definido). Eixo X = score de risco (PD*100).
e = train[['PD','default']].copy()
e['score_risco'] = (e['PD']*100).clip(0,60)
alvo = 12.0          # teto de inadimplência acumulada aceito (decisão de negócio)
step = 2
fx = np.arange(0, 60, step)
reg = []
for b in fx:
    aprov = e[e['score_risco'] <= b+step]
    faixa = e[(e['score_risco']>b)&(e['score_risco']<=b+step)]
    reg.append({'banda':b,
                'pop_acum': len(aprov)/len(e)*100,
                'bad_acum': aprov['default'].mean()*100 if len(aprov) else np.nan,
                'bad_faixa': faixa['default'].mean()*100 if len(faixa) else np.nan})
reg = pd.DataFrame(reg)
corte_risco = CORTE_RISCO
row = reg.loc[reg['banda']==corte_risco-step].iloc[0]

fig, ax = plt.subplots(figsize=(13,5.6))
ax.bar(fx, reg['pop_acum'], width=step*0.82, color='#D9D9D9', label='Pop Aprovada (acum.)', zorder=1, align='edge')
xc = fx + step/2
ax.plot(xc, reg['bad_faixa'], color='#222', lw=2, marker='o', ms=4, label='Inadimplência da faixa', zorder=3)
ax.plot(xc, reg['bad_acum'], color='#C0392B', lw=2, marker='o', ms=4, label='Inadimplência acum. (aprovados)', zorder=3)
ax.axhline(alvo, color='#2E86C1', ls=':', lw=1.6, label=f'Inadimplência máxima aceita ({alvo:.0f}%)')
ax.axvline(corte_risco, color='#333', ls='--', lw=1.4)
ax.annotate(f'Corte PD <= {corte_risco:.0f}%\naprova {row.pop_acum:.0f}% do treino\ninadimplência {row.bad_acum:.1f}%',
            xy=(corte_risco,alvo), xytext=(corte_risco+3,alvo+16), fontsize=11,
            bbox=dict(boxstyle='round,pad=0.5',fc='#FFF7D6',ec='#D4AC0D'),
            arrowprops=dict(arrowstyle='->',color='#333'))
for k,(xb,v) in enumerate(zip(xc,reg['pop_acum'])):
    if pd.notna(v) and k%2==0: ax.text(xb,v+1.8,f'{v:.0f}',ha='center',fontsize=8.5,color='#555')
ax.set_title('Strategy curve no treino — definição do ponto de corte', fontsize=14, fontweight='bold', color='#1B2A4A', pad=12)
ax.set_xlabel('Score de risco = PD (%) — ponto de corte', fontsize=12)
ax.set_ylabel('%', fontsize=12); ax.set_ylim(0,112); ax.set_xticks(np.arange(0,62,4))
ax.legend(loc='center left', bbox_to_anchor=(1.01,0.5), frameon=False, fontsize=10)
ax.spines[['top','right']].set_visible(False)
plt.tight_layout(); plt.show()
```

Aplicado ao conjunto de teste (validação out-of-sample), o corte PD ≤ 20% aprova 33.332 dos 71.206 contratos (46,8%), com inadimplência de 14,56% entre os aprovados, contra 36,73% entre os reprovados e 26,35% na base de teste. A carteira aprovada inadimple a pouco mais da metade da taxa do pool de solicitantes, o que confirma que o modelo ordena adequadamente o risco e fornece uma base sólida para a etapa de otimização.

Cabe notar que a inadimplência dos aprovados no teste (14,56%) fica acima da calibrada no treino (11,86%), porque o período de teste é mais arriscado; é o desempenho honesto fora da amostra.
